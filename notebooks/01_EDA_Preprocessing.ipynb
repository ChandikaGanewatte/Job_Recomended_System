This notebook focuses on data exploration and preprocessing of candidate and job datasets. The goal is to clean, transform, and prepare the data for building recommendation models.

🔹 Key Steps:
    1.Load &Sample
    2.Basic Cleaning
    3.Fast Text Cleaning
    4.Save processed data 

🔹 Output:
    processed_resume.csv
    processed_jobs.csv

1. Load & Sample

```python
import pandas as pd
import numpy as np

resume_df = pd.read_csv('../data/raw/resume_dataset_200k_enhanced.csv')
jobs_df   = pd.read_csv('../data/raw/postings.csv')

# 🔥 Sample for speed
resume_df = resume_df.sample(20000, random_state=42)
jobs_df   = jobs_df.sample(5000, random_state=42)
```

2. Basic Cleaning

```python
resume_df.drop_duplicates(inplace=True)
jobs_df.drop_duplicates(inplace=True)

# Handle missing values safely
for df in [resume_df, jobs_df]:
    text_cols = df.select_dtypes(include=['object']).columns
    num_cols  = df.select_dtypes(exclude=['object']).columns

    df[text_cols] = df[text_cols].fillna('')
    df[num_cols]  = df[num_cols].fillna(df[num_cols].median())
```

3. Fast Text Cleaning

```python
def clean_text(series):
    return (series.str.lower()
            .str.replace(r'http\S+', '', regex=True)
            .str.replace(r'[^\w\s]', '', regex=True)
            .str.replace(r'\s+', ' ', regex=True)
            .str.strip())

resume_df['text'] = clean_text(
    resume_df['education_level'] + " " +
    resume_df['university_tier'] + " " +
    resume_df['company_type']
)

jobs_df['text'] = clean_text(
    jobs_df['title'] + " " +
    jobs_df['description'] + " " +
    jobs_df['skills_desc']
)
```

4. Save Processed Data

```python
resume_df.to_csv('../data/processed/processed_resume.csv', index=False)
jobs_df.to_csv('../data/processed/processed_jobs.csv', index=False)
```
